# 1. Configuración y símbolos

Los símbolos $x$, $y$, $p$ y $q$ se declaran sobre $\mathbb{R}$ mediante la hipótesis `real=True`.

```python
import sympy as sp
from IPython.display import display

sp.init_printing(use_latex='mathjax')

x, y, p, q = sp.symbols('x y p q', real=True)

display(sp.Tuple(x, y, p, q))
```

# 2. Sustitución de valores (`subs`)

La sustitución simple reemplaza un símbolo por un valor en una expresión. La sustitución múltiple aplica simultáneamente un diccionario de pares símbolo–valor. Para una expresión $f(x,y)$, se usan `f.subs(x, valor)` y `f.subs({x: valor_x, y: valor_y})`, respectivamente.

```python
U = 2 * x**sp.Rational(1, 2) * y**sp.Rational(1, 2)

display(sp.Eq(sp.Function('U')(x, y), U))
```

```python
U_x_4 = U.subs(x, 4)

display(U_x_4)
```

```python
U_4_9 = U.subs({x: 4, y: 9})

display(U_4_9)
```

# 3. Resultados exactos vs. aproximados (`Rational`, `evalf`)

Un literal de Python como $0.5$ se representa mediante punto flotante binario y puede introducir aproximaciones antes de la simplificación simbólica. Para conservar exponentes y coeficientes racionales exactos, se usa `sp.Rational(1, 2)` en lugar de $0.5$. La conversión numérica debe posponerse hasta el final con `.evalf(n)`, donde $n$ controla la precisión solicitada.

```python
expresion_exacta = sp.sqrt(2) / 10 + sp.Rational(1, 7)

display(expresion_exacta)
```

```python
aproximacion_5 = expresion_exacta.evalf(5)

display(aproximacion_5)
```

# 4. Resolución de ecuaciones (`sp.Eq`, `solve` / `linsolve`)

En Python, `=` asigna un objeto a un nombre; no define una igualdad algebraica. En SymPy, `sp.Eq(a, b)` construye la relación matemática $a=b$ que puede entregarse a un solucionador.

Para el equilibrio de mercado:

$$
Q_d(p)=100-2p,
\qquad
Q_s(p)=20+3p.
$$

El precio de equilibrio $p^*$ satisface $Q_d(p^*)=Q_s(p^*)$.

```python
Q_d = 100 - 2 * p
Q_s = 20 + 3 * p

ecuacion_equilibrio = sp.Eq(Q_d, Q_s)
solucion_p = sp.solve(ecuacion_equilibrio, p, dict=True)

display(ecuacion_equilibrio)
display(solucion_p)
```

# 5. Ecuaciones no lineales (`nonlinsolve` / `solve`)

Para una función objetivo no lineal, los candidatos interiores se obtienen buscando las raíces de la condición de primer orden $\pi'(q)=0$. En este caso:

$$
\pi'(q)=-4q+50.
$$

```python
pi_prime = -4 * q + 50
condicion_primer_orden = sp.Eq(pi_prime, 0)
solucion_q = sp.solve(condicion_primer_orden, q)

display(condicion_primer_orden)
display(solucion_q)
```

# 6. Pipeline integrado: resolver y sustituir

Con `solve(..., dict=True)`, el resultado es una lista de diccionarios. El valor $p^*$ se extrae con `solucion_p[0][p]` y se inyecta automáticamente en $Q_d(p)$ mediante `.subs()`:

$$
q^*=Q_d(p^*).
$$

```python
p_star = solucion_p[0][p]
q_star = sp.simplify(Q_d.subs(p, p_star))

display(sp.Eq(sp.Symbol('p^*'), p_star))
display(sp.Eq(sp.Symbol('q^*'), q_star))
```
